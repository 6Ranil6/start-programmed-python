```python
import numpy as np
x = np.array([1, 2, 3, 4])
y = np.array([9, 2, 3, 4])
k = 5
print(x + y) # поэлементное сложение
print(k * x)
print(x.tolist())
print(x.T) # для одномерной матрицы ничего не меняется 
print(np.array([[1],
               [2],
               [3]])) # имеет разницу

x = np.array([x.tolist(),y.tolist()]) # надо ставить [], т.к. внутри массива у нас два подмассива

print(x.transpose())
print(x.T)
```

```python
from numpy import linalg as l 
x = np.array([1, 2, 3])
print(x)
y = np.array([2, 3, 4])
print(y)

print(l.norm(x)) # находит длину одного вектора

print(np.sqrt(((x - y) ** 2).sum()))# находит растояние между двумя точками (векторами идущих из начала координат)
print(l.norm((x - y))) # аналогично


print(np.power(x, 2))
print(np.power(x, 2).sum())
print(np.sqrt(np.power(x, 2).sum()))
```

```python
x = np.array([1, 2, 3])
y = np.array([2, 3, 4])
print(x * y) # поэлементное произведение

print(np.outer(x, y)) # произведени 2-х матриц

# !!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!

print(x @ y) # скалярное произведение
print(np.dot(x, y)) # сколярное произведение
print(np.inner(x, y))
```

```python
# посчитаем косинус между векторами
cosa = x @ y / (l.norm(x) * l.norm(y))

print(cosa)
theta_rad = np.arccos(cosa) # возвращает угл в радианах
print(theta_rad)
a = theta_rad * 180 / np.pi
print(a)
```
