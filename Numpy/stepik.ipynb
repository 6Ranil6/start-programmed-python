```python
import pandas as pd

print(pd.Series(5))

print(pd.Series(5, index=[1, 2, 3]))

print(pd.Series(5, index=['a', 'b', 'c']))

print(pd.Series(5, index= {1, 2, 3}))

print(pd.Series(5, index= (1, 2, 3)))

print(pd.Series([(1, 'a'), (2, 'b')]))

print(pd.Series({1:'a', 'b': 2}))
```
