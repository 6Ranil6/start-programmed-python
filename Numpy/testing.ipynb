Создание двумерного массива

Создайте двумерный массив (матрицу) размером 4x4, заполненный случайными целыми числами от 1 до 100.

```python
import numpy as np 

ndarray = np.array([], ndmin= 2)
A = np.random.randint(1, 100, size= (4, 2))
B = np.random.randint(1, 100, size= (4, 2)) 
print(A, B)
ndarray = np.append(A, B, axis= 1)
ndarray
```

Извлеките первый и третий ряды (строки) из матрицы.

Извлеките второй и четвертый столбцы.

```python
print(ndarray[:,[0,2]])
print(ndarray[(1,3),:])
```

Привести массив к другому типу

```python
ndarray.astype('float64')
```

Увеличьте все элементы матрицы на 10.

Замените элемент в третьей строке и втором столбце на значение 0.

```python
print(ndarray + 10)
ndarray[2,1] = 0 
print(ndarray)
```

```python
import numpy as np

A = np.random.randint(1, 11, size= (1, 10))
B = np.random.randint(1, 11, size= (1, 10))
C = np.random.randint(1, 11, size= (1, 10))

print(f"A:\n {A}")

print(f"B:\n {B}")

print(f"C:\n {C}")

print("Concate:\n", np.concatenate((A, B), axis= 0))

print("Vstack:\n", np.vstack([B,C]))

print('Hstack:\n', np.hstack([A, C]))

print("Np_c:",  np.c_[B[:, :3], C])

print("np_r:", np.r_[A, C])
```

```python
# Диапазон значений
F = np.arange(stop= 5, step= 0.5) # шаг мы выставляем 
print(F)
# 
G = np.linspace(start=0, stop= 5, num=  20) # шаг сам высчитыввает
print(G)
E = np.linspace(start=0, stop= 5, num=  10, endpoint= True, retstep= True) # шаг сам высчитыввает
print(E)
B = np.logspace(
    start= 7, # Стартовая степень
    stop= 10, # Конечная степень
    num= 5, 
    endpoint= False,
    base= 2) # в основании 2 
print(B)
```

```python
np.random.seed(1)
X = np.random.randint(1, 10, size= (3, 3))
print(X)
print(X[(X < 5) & (X > 1)]) # выбор элементов по условию
print(X[ # выбор по индексу
    [0, 1, 2],  # координаты по X
    [0, 1, 2] # координаты по Y
])
```

# Флаги

```python
C = np.random.randint(0, 2, size= (3, 3), dtype= 'bool') # Для типа bool start = 0, stop = 2
print(X)
print(C)
print(X[C]) # УБИРАЕТ элементы с позиции False
```

```python
print(X.sum(axis= 1)) # сложение по "строчкам" axis = 1
print(X.sum(axis= 0))  # сложение по "столбцам" axis = 0

print(X[X.sum(axis= 1) > 10]) # вывод строк, где сумма элементов больше 10
print(X[[0, 2]])
X[:,[0, 2]]
```

```python
print(X)
print(np.where(X < 5)) # возвращает координаты по X и по Y элементов вернувших True

print(X[X<5]) # Возвращает сами элементы

print(np.where(X < 5, X, -1))
```

```python
%%timeit
def add_2(x):
    return x + 2

new_x = np.vectorize(add_2) # делает функцию применимой для ndarray
new_x(X)
```

```python
%%timeit
from numba import vectorize
@vectorize
def add_2(x):
    return x + 2

new_x = add_2(X)
print(new_x)
```
