Pandas

Рассчитывать среднее, медиану и моду мы можем отдельно для каждого столбца датафрейма Pandas, например, для переменной рост игрока (height):

# Среднее
mean_height = df['height'].mean()  # на выходе число

# Медиана
median_height = df['height'].median()  # на выходе число

# Мода
mode_height = df['height'].mode()  # на выходе серия Pandas со значениями мод
А можем получить описательные статистики по всем столбцам с числовыми данными, применив методы .mean(), .median() и .mode() сразу ко всему датафрейму:

mean = df.mean()
print(mean)

>>
player_number     13.000000
height           178.545455
weight            76.727273
score             90.181818
dtype: float64
Важно понимать, что расчет среднего, медианы и моды не имеет смысла для таких значений, как номер игроков.

Применять эти методы можно и к данным, запакованным в отдельные серии (pd.Series), например:

weight = pd.Series([76, 77, 79, 81, 82, 79, 74, 84, 73, 71, 68])

mean_weight = weight.mean()
print(mean_weight)

>> 76.72727272727273

from scipy import stats

# Среднее
mean_height = np.mean(df['height'])

```python
import pandas as pd
import numpy as np
from scipy import stats
data = {"player_number": np.array([10, 8, 12, 22, 36, 7, 1, 20, 9, 4, 14]),
        "height": np.array([168, 176, 178, 191, 185, 183, 185, 179, 169, 183, 167]),
        "weight": np.array([76, 77, 79, 81, 82, 79, 74, 84, 73, 72, 68]),
        "score": np.array([95, 86, 94, 96, 95, 95, 89, 83, 99, 78, 82])
}
df = pd.DataFrame(data)
print(round(df['score'].median(), 1))
A = df
print(round(A['weight'].mean(), 5))

print(round(A["weight"][A["score"] > 90].mean(), 5))

print(round(A['height'][(A['weight'] > 80) & (A["score"] >= 95)].mean(), 1))

print(A["score"].mode().iloc[0].round(0))
```
