# Бинарный поиск O(log(n)): 

Делить <u>отсортировнный</u> массив на две части

Реализация:

```python
from typing import Iterable
def binary_search(data: Iterable, item) -> int:
    """
    :param data: - итерируемый объект, по которому производится поиск
    :param item: - элемент, который ищем
    :return index: индекс найденного элемента или -1, если элемент не найден
    :rtype int:
    """
    top = 0
    end = len(data) - 1
    while end >= top:
        mid = (end + top) // 2
        if data[mid] == item:
            return mid
        elif data[mid] > item:
            end = mid - 1
        else:
            top = mid + 1 
    return None
            

print(binary_search([i for i in range(100) if i > 10], 98))
```
