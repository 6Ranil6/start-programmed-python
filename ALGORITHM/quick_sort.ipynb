Быстрая сортировка

Идея заключается в том, что есть один опорный элемент, через который происходит деление на 2 подмассива.

Базовые случаи - это 0 и 1 элемент в массиве

```python
from random import randint

def quick_sort(arr: list):
    
    if len(arr) <= 1:
        print(arr)
        return arr

    pivot_el = arr[randint(0, len(arr) - 1)]

    left_els = []
    right_els = []
    pivot = []

    for el in arr:
        if el < pivot_el:
            left_els.append(el)
        elif el > pivot_el:
            right_els.append(el)
        else:
            pivot.append(el)
            
    return quick_sort(left_els) + pivot + quick_sort(right_els)

print(quick_sort([10, 3, 9, 4, 3]))

    
```
